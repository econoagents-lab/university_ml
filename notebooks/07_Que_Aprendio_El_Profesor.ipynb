## 6. Visualizaciones

Cada gráfico debe responder una pregunta de negocio. No graficamos por decoración.

```python
df = load_operations()
fig, ax = plt.subplots(figsize=(10, 5))
summary = df.groupby("canal_agrupado")["caida_30d"].mean().sort_values()
ax.bar(summary.index, summary.values)
ax.set_title("Tasa de caída por canal")
ax.set_xlabel("Canal")
ax.set_ylabel("Tasa de caída 30d")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.scatter(df["dias_en_tuberia"], df["precio_departamento"], alpha=0.25)
ax.set_title("Relación entre días en tubería y precio")
ax.set_xlabel("Días en tubería")
ax.set_ylabel("Precio departamento")
plt.show()
```
